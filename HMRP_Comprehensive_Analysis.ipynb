## ⚙️ Step 1: Environment Setup & Module Imports

```python
import os
import sys
from pathlib import Path
import pandas as pd
import plotly.io as pio

# Detect environment
IN_COLAB = "google.colab" in sys.modules
pio.renderers.default = "colab" if IN_COLAB else "notebook_connected"

PROJECT_ROOT = Path(os.getcwd()).resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from app.pipeline import AnalysisPipeline
from visualization.hierarchy_viz import HierarchyViz
from visualization.knowledge_graph import KnowledgeGraph
from visualization.mindmap_viz import MindMapViz
from visualization.heatmap_viz import HeatmapViz
from hierarchy.ccc_mapper import CCCMapper
from models.paper import Paper

print("✅ All HMRP modules & visualization classes imported successfully!")
```

## 📥 Step 2: Ingest and Process Multidisciplinary Papers

```python
pipeline = AnalysisPipeline()
ccc_mapper = CCCMapper()

pdf_files = list(Path("HMRP_Dataset/PDFs").rglob("*.pdf"))
print(f"[*] Discovered {len(pdf_files)} local PDFs on disk.")

if len(pdf_files) >= 3:
    papers = pipeline.process_papers(pdf_files[:6])
    print(f"✅ Successfully processed {len(papers)} papers from disk.")
else:
    print("[*] Generating Multidisciplinary Benchmark Papers spanning 5 domains...")
    p1 = Paper(paper_id="demo_nlp", title="Transformer Architectures for Hierarchical Concept Mapping", file_name="nlp_transformer.pdf", page_count=14, word_count=9800, abstract="We propose multi-head self-attention transformers for hierarchical concept taxonomy generation.", research_domains=["Natural Language Processing", "Artificial Intelligence"], research_concepts=["Transformer", "Self-Attention", "Concept Hierarchy", "Language Models", "Semantic Parsing"], all_keywords=["transformer", "nlp", "attention mechanism", "deep learning"])
    p2 = Paper(paper_id="demo_robotics", title="Real-Time 3D Point Tracking with State Space Models in Robotics", file_name="robotics_ssm.pdf", page_count=18, word_count=14200, abstract="This paper introduces state space models for 3D point tracking and robotic obstacle avoidance.", research_domains=["Robotics", "Computer Vision", "Artificial Intelligence"], research_concepts=["State Space Models", "Point Tracking", "Robotic Manipulation", "Computer Vision", "Deep Learning"], all_keywords=["robotics", "point tracking", "state space model", "reinforcement learning"])
    p3 = Paper(paper_id="demo_bio", title="Deep Learning for 3D Protein Structure Prediction from Genomic Sequences", file_name="genomics_protein.pdf", page_count=16, word_count=11500, abstract="We apply transformer sequence encoders to predict 3D molecular protein structures from genomic sequences.", research_domains=["Biomedical & Genomics", "Artificial Intelligence"], research_concepts=["Transformer", "Protein Structure Prediction", "Genomics", "Amino Acids", "Deep Learning"], all_keywords=["protein folding", "transformer", "genomics", "bioinformatics"])
    p4 = Paper(paper_id="demo_quantum", title="Quantum Entanglement Verification via Optical Interferometry", file_name="quantum_optics.pdf", page_count=15, word_count=10200, abstract="We formulate quantum state tomography and verify photonic entanglement using interferometers.", research_domains=["Quantum Physics & Optics", "Applied Mathematics & Optimization"], research_concepts=["Quantum Entanglement", "Photonic Qubits", "Interferometry", "State Tomography"], all_keywords=["quantum optics", "entanglement", "qubits", "photonics"])
    p5 = Paper(paper_id="demo_finance", title="Stochastic Volatility Modeling and Portfolio Optimization in Quantitative Finance", file_name="quant_finance.pdf", page_count=20, word_count=13800, abstract="We formulate convex optimization and Black-Scholes risk models for high-frequency algorithmic portfolio rebalancing.", research_domains=["Quantitative Finance", "Applied Mathematics & Optimization"], research_concepts=["Portfolio Optimization", "Black-Scholes Model", "Stochastic Volatility", "Convex Optimization"], all_keywords=["quantitative finance", "portfolio optimization", "risk management", "convex optimization"])
    papers = [p1, p2, p3, p4, p5]
    for p in papers:
        tree, _ = pipeline._hierarchy.build(p.research_concepts, p.all_keywords, p.abstract)
        p.concept_hierarchy = tree
        p.mindmap = {"root": p.title, "children": tree}
        p.ccc_mapping = ccc_mapper.map_paper_ccc(p.research_concepts, p.all_keywords, p.research_domains)
print(f"✅ Prepared {len(papers)} papers.")
```

## 📊 Step 3: Hierarchical Concept Taxonomy Visualizations

```python
hier_viz = HierarchyViz()
selected_paper = papers[0]
print(f"📌 Visualizing Hierarchy for: {selected_paper.title}")

fig_sunburst = hier_viz.sunburst(selected_paper.concept_hierarchy, title=f"Sunburst Hierarchy: {selected_paper.title[:50]}")
fig_sunburst.show()

fig_treemap = hier_viz.treemap(selected_paper.concept_hierarchy, title=f"Treemap Hierarchy: {selected_paper.title[:50]}")
fig_treemap.show()

fig_icicle = hier_viz.icicle(selected_paper.concept_hierarchy, title=f"Icicle Hierarchy: {selected_paper.title[:50]}")
fig_icicle.show()
```

## 🧠 Step 4: Radial Mind Map Visualization

```python
mindmap_viz = MindMapViz()
fig_mindmap = mindmap_viz.build(title=selected_paper.title[:35], tree=selected_paper.concept_hierarchy)
fig_mindmap.show()
```

## 🕸️ Step 5: Concept-to-Concept (CCC) Semantic Network & Cross-Domain Bridges

```python
kg = KnowledgeGraph()
ccc_info = getattr(selected_paper, "ccc_mapping", {})

if ccc_info:
    kg.build_from_ccc(ccc_info, paper_title=selected_paper.title[:45])
    fig_ccc = kg.to_plotly(title=f"CCC Semantic Network: {selected_paper.title[:45]}")
    fig_ccc.show()
    print("🌉 Detected Cross-Domain Bridge Concepts:")
    for b in ccc_info.get("cross_domain_bridges", []):
        print(f"  • {b.get('concept')} (Spans: {', '.join(b.get('connected_domains', []))})")
    print("\n🔗 Top Concept-to-Concept Associations:")
    for e in ccc_info.get("c2c_edges", [])[:8]:
        print(f"  • {e.get('concept_a')} <-> {e.get('concept_b')} (Weight: {e.get('weight', 0):.2f})")
else:
    print("No CCC mapping available.")
```

## 📈 Step 6: Pairwise Multidisciplinary Similarity Heatmaps

```python
sim_results = pipeline.compute_similarity(papers, method="both")
heatmap_viz = HeatmapViz()
paper_labels = [p.title[:30] for p in papers]

if sim_results.get("combined_matrix") is not None:
    fig_heatmap = heatmap_viz.build(sim_results["combined_matrix"], labels=paper_labels, title="Multidisciplinary Inter-Paper Combined Similarity Heatmap")
    fig_heatmap.show()
```

## 🌐 Step 7: Global Corpus Multi-Domain CCC Mapping & Bridge Ranking

```python
corpus_ccc = ccc_mapper.map_corpus_ccc(papers)
print("=" * 75)
print(f"GLOBAL CORPUS MULTIDISCIPLINARY CCC ANALYSIS ({len(papers)} Papers)")
print("=" * 75)
print(f"Total Unique Concepts Tracked: {corpus_ccc.get('total_unique_concepts', 0)}")
print("\n🌟 Top Cross-Domain Bridge Concepts:")
for b in corpus_ccc.get("cross_domain_bridge_concepts", [])[:10]:
    print(f"  • {b.get('concept', ''):<32} | Frequency: {b.get('frequency', 0)} | Domains: {', '.join(b.get('domains', []))}")
print("\n🔗 Top Co-occurring Concept Pairs:")
for pair in corpus_ccc.get("top_cooccurring_concept_pairs", [])[:8]:
    print(f"  • {pair.get('concept_a', '')} <---> {pair.get('concept_b', '')} (Co-occurrence Count: {pair.get('cooccurrence_count', 0)})")
print("=" * 75)
```
